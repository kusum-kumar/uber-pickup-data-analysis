```python
import pandas as pd
import plotly.express as px

# Check the original data
print("Original data shape:", uber_15.shape)
print(uber_15["Pickup_date"].head())

# Convert pickup dates
uber_15["Pickup_date"] = pd.to_datetime(
    uber_15["Pickup_date"], errors="coerce"
)

# Remove rows with invalid dates
uber_15 = uber_15.dropna(subset=["Pickup_date"]).copy()

print("Valid rows:", len(uber_15))

# Extract month and weekday
uber_15["month"] = uber_15["Pickup_date"].dt.month
uber_15["weekday"] = uber_15["Pickup_date"].dt.day_name()

# Create and sort the pivot table
month_order = list(range(1, 13))
weekday_order = [
    "Monday", "Tuesday", "Wednesday", "Thursday",
    "Friday", "Saturday", "Sunday"
]

pivot_sorted = pd.crosstab(
    uber_15["month"], uber_15["weekday"]
).reindex(
    index=month_order, columns=weekday_order, fill_value=0
)

print(pivot_sorted)

# Convert the pivot table into chart-friendly format
chart_data = (
    pivot_sorted.rename_axis("month")
    .reset_index()
    .melt(
        id_vars="month",
        var_name="weekday",
        value_name="pickups"
    )
)

# Create the chart
fig = px.bar(
    chart_data,
    x="month",
    y="pickups",
    color="weekday",
    barmode="group",
    title="Uber pickups by Month & weekday",
    labels={
        "month": "Month",
        "pickups": "Number of pickups",
        "weekday": "Weekday"
    }
)

fig.show()
```

```python
import pandas as pd
import plotly.express as px

# Extract month and weekday
uber_15["Pickup_date"] = pd.to_datetime(uber_15["Pickup_date"])
uber_15["month"] = uber_15["Pickup_date"].dt.month
uber_15["weekday"] = uber_15["Pickup_date"].dt.day_name()

# Create pivot table
pivot = pd.crosstab(
    index=uber_15["month"],
    columns=uber_15["weekday"]
)

# Sort months and weekdays
month_order = [1, 2, 3, 4, 5, 6]
weekday_order = [
    "Monday", "Tuesday", "Wednesday", "Thursday",
    "Friday", "Saturday", "Sunday"
]

pivot_sorted = pivot.reindex(
    index=month_order,
    columns=weekday_order,
    fill_value=0
)

# Create chart data
chart_data = pivot_sorted.reset_index().melt(
    id_vars="month",
    var_name="weekday",
    value_name="pickups"
)

# Plot the chart
fig = px.bar(
    chart_data,
    x="month",
    y="pickups",
    color="weekday",
    barmode="group",
    title="Uber pickups by Month & weekday",
    labels={
        "month": "Month",
        "pickups": "Number of pickups",
        "weekday": "Weekday"
    },
    category_orders={
        "weekday": weekday_order,
        "month": month_order
    }
)

fig.update_layout(
    xaxis=dict(
        tickmode="array",
        tickvals=month_order,
        ticktext=[str(m) for m in month_order]
    ),
    yaxis=dict(
        title="Number of pickups",
        rangemode="tozero"
    )
)

fig.show()
```

```python
import pandas as pd

files = [
    "uber-raw-data-apr14.csv",
    "uber-raw-data-may14.csv",
    "uber-raw-data-jun14.csv",
    "uber-raw-data-jul14.csv",
    "uber-raw-data-aug14.csv",
    "uber-raw-data-sep14.csv"
]

dataframes = []

for file in files:
    df = pd.read_csv(file)
    dataframes.append(df)
    print(file, df.shape)

final = pd.concat(dataframes, ignore_index=True)

print("Final shape:", final.shape)
print(final.head())
```

```python
import pandas as pd

final["Date/Time"] = pd.to_datetime(final["Date/Time"], errors="coerce")
final["Lat"] = pd.to_numeric(final["Lat"], errors="coerce")
final["Lon"] = pd.to_numeric(final["Lon"], errors="coerce")

final["hour"] = final["Date/Time"].dt.hour

clean_data = final.dropna(subset=["Lat", "Lon", "hour"]).copy()

clean_data = clean_data[
    clean_data["Lat"].between(40, 42) &
    clean_data["Lon"].between(-75, -72)
]

hourly_data = []

for h in range(24):
    temp = clean_data.loc[
        clean_data["hour"] == h, ["Lat", "Lon"]
    ]
    hourly_data.append(temp.values.tolist())

print("Valid records:", len(clean_data))
print("Hourly groups:", len(hourly_data))
print("Records in first hour:", len(hourly_data[0]))
```

```python
import folium
from folium.plugins import HeatMapWithTime
from IPython.display import display

m = folium.Map(
    location=[40.75, -73.97],
    zoom_start=11,
    tiles="CartoDB positron"
)

HeatMapWithTime(
    hourly_data,
    index=[f"{h}:00" for h in range(24)],
    radius=8,
    auto_play=False,
    max_opacity=0.8
).add_to(m)

m.save("uber_hourly_heatmap.html")

display(m)
```
